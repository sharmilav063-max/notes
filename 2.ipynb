```python
#create a function to print prime numbers between given range using all 4 methods 

def prime1():
    a = int(input("Enter starting number: "))
    b = int(input("Enter ending number: "))

    for i in range(a, b+1,1):
        if (i==1):
            continue
     
        for j in range(2, i,1):
            if i % j == 0:
                break

        else:
            print(i)
```

```python
prime1()
```

```python
#create a function to print prime numbers between given range using all 4 methods 

def prime2():
    pn=[]
    a = int(input("Enter starting number: "))
    b = int(input("Enter ending number: "))

    for i in range(a, b+1,1):
        if (i==1):
            continue
     
        for j in range(2, i,1):
            if i % j == 0:
                break

        else:
            pn.append(i)
    return pn


            


            
```

```python
prime2()
```

```python
for i in prime2():
    print(i)
```
