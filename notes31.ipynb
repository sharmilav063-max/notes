```python
for i in range(1,6,1):
    for j in range(1,6,1):
        if (i*j)%2==0:
            print(i*j, end="\t")
        else:
            print(" ",end ="\t")
    print("\n")
```

```python
syntax of while loop inside while loop

initalization of outer while
while(outer condition):

    initalization of inner while
    while (inner condition):
        statement of inner while
        inc/dec of inner while

    statemnt of outer while
    inc/dec of inner while
```

```python
i=1
while(i<=5):
    j=1
    while(j<=5):
        print("*",end=" ")
        j+=1

    print()
    i+=1
```
