```python
#calculate sum of digit as a user input(#123=6)

num= int(input("enter a number:")) #initalization
s=0
while (num>0):
    rem=num%10
    s=s+rem
    num=num//10
print("the sum of digits is:",s)

           
```

```python
# write a program palindrome for number

num=int(input("enter a number:"))
print("the number before reverse:",num)
temp=num
rev=0
while num>0:
    rem = num%10
    rev=(rev*10)+rem
    num =num//10
print("the number after reverse:",rev)
if(temp==rev):
    print("the number is palindrome")
else:
    print("the number is not palindrome")
```
